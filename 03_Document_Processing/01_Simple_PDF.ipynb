## 1. Simple PDF Loaders

- **Loader:** `PyPDFLoader`
- **Install:** `pip install pypdf`
- **Use Case:** Standard text-heavy PDFs with single-column layouts (reports, simple essays).

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader = PyPDFLoader("../docs/document_processing/simple-text.pdf")
docs = loader.load() # Extracts the PDF content and returns one Document object per page. The text is split naturally at page boundaries, keeping each page intact.
```

```python
print(f"Total Documents {len(docs)}")
for i, doc in  enumerate(docs):
    print(f"\n\nPage No: {i+1}")
    print(f"Source: {doc.metadata.get("source", "None")}")
    print(f"Content: {doc.page_content[:40]}....{doc.page_content[-40:]}")
```
