### 5. Markdown (`.md`)

* **Loader:** `UnstructuredMarkdownLoader`

* **Install:** `pip install unstructured markdown`

* **Use Case:** Technical documentation; preserves header hierarchy (`#`, `##`, `###`).


```python
from langchain_community.document_loaders import UnstructuredMarkdownLoader
```

```python
loader = UnstructuredMarkdownLoader('../docs/document_processing/sample.md')
docs = loader.load()
```

```python
print(f"Total Documents: {len(docs)}")
for i, doc in enumerate(docs):
    print(f"\n\nPage No: {i+1}")
    print(f"Source: {doc.metadata.get("source", "None")}")
    print(f"Content: {doc.page_content[:40]}....{doc.page_content[-40:]}")
```
