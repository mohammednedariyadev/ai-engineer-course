# Matplotlib — ⭐⭐⭐⭐

## Setup
Every script begins with importing the `pyplot` interface:

```python
import numpy as np
import matplotlib.pyplot as plt
```

### 1. `plt.plot()` — Line Chart

Used for continuous data, trends over time, or mathematical curves. Points are connected by straight lines.

**Best for:** Time series, sequential measurements, and functions **$y = f(x)$**.

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 25, 20, 35, 30])

plt.plot(
    x, y,
    color="tab:blue",       # Line color
    linestyle="--",         # Line Style: '-', '--', '-.', ':'
    linewidth=2,            # Line thickness
    marker="o",             # Point symbols: 'o', 's', '^', etc.
    markersize=6,           # Size of markers
    label="Revenue Trend"    # Label for legend
)

plt.xlabel("Month")
plt.ylabel("Revenue ($K)")
plt.title("Monthly Revenue Growth")
plt.legend()
plt.show()
```

### 2. `plt.scatter()` — Scatter Plot
Used to plot individual data points without connecting lines to reveal relationships, correlations, clusters, or outliers between two numeric variables.

- **Best for:** Bivariate relationships (e.g., height vs. weight, price vs. demand).

```python
study_hours = np.array([2, 4, 5, 7, 8, 3, 9, 6])
exam_scores = np.array([55, 68, 72, 85, 88, 60, 95, 78])

plt.scatter(
    study_hours,
    exam_scores,
    color="crimson",
    marker="o",
    s=80,               #Size of markers (can be an array to make bubble charts)
    alpha=0.8,          # Opacity (0 = transparent, 1 = opaque; helps with overlapping points)
    edgecolors="black"
)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score (%)")
plt.title("Study Time vs. Exam Performance")
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

### 3. `plt.bar()` — Bar Chart

Used to compare discrete categorical groups against a single numeric measurement.

- *Tip:* For horizontal bars (useful for long category names), use `plt.barh(categories, sales)`.

```python
categories = ["Product A", "Product B", "Product C", "Product D"]
sales = [120, 240, 180, 95]

plt.bar(
    categories,
    sales,
    color=["#4C72B0", "#55A868", "#C44E52", "#8172B3"],
    width=0.6,      # Bar width (default 0.8)
    edgecolor="black"
)

plt.xlabel("Product Category")
plt.ylabel("Units Sold")
plt.title("Q1 Product Sales Comparison")
plt.show()
```

### 4. `plt.hist()` — Histogram
Used to show the distribution of a single continuous variable by grouping values into adjacent intervals called "bins" and counting the frequency per bin.

- *Best for:* Checking skewness, unimodal/bimodal patterns, and normality in a single feature.

```python
# Generate 500 values centered around mean 170 with std 10
heights = np.random.normal(loc=170, scale=10, size=500)

plt.hist(
    heights,
    bins=15,                # Number of equal-width bins (or pass custom bin edges)
    color="skyblue",
    edgecolor="black",      # Distinguishes individual bin bars
    density=False           # True gives a probability density (area sums to 1)
)

plt.xlabel("Height (cm)")
plt.ylabel("Frequency (Count)")
plt.title("Distribution of Sample Heights")
plt.show()
```

### 5. `plt.boxplot()` — Box and Whisker Plot
Used to visualize the five-number summary: **minimum, first quartile ($Q_1$), median, third quartile ($Q_3$), and maximum**, along with statistical outliers.

- **How to read it:**
    - **Center line:** Median ($Q_2$).
    - **Box edges:** 25th percentile ($Q_1$) to 75th percentile ($Q_3$) — the Interquartile Range (IQR).
    - **Whiskers:** Reach up to $1.5 \times \text{IQR}$ beyond the quartiles.
    - **Points past whiskers:** Flagged as potential outliers.

```python
# Create sample score distributions for 3 classes
class_a = [45, 55, 60, 62, 65, 70, 72, 75, 80, 98]  # 98 is an outlier
class_b = [60, 65, 68, 70, 72, 74, 76, 78, 82, 85]
class_c = [50, 52, 55, 58, 60, 61, 63, 65, 68, 70]

plt.boxplot(
    [class_a, class_b, class_c],
    tick_labels=["Class A", "Class B", "Class C"],
    vert=True,            # Set to False for horizontal layout
    patch_artist=True,    # Fills boxes with color
    boxprops=dict(facecolor="lightyellow", color="black"),
    flierprops=dict(marker="o", markerfacecolor="red", markersize=6) # Outlier style
)

plt.ylabel("Exam Score")
plt.title("Score Spread and Outliers Across Classes")
plt.show()
```
