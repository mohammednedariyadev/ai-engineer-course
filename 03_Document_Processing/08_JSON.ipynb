### 8. JSON / JSONL

* **Loader:** `JSONLoader`

* **Install:** `pip install jq`

* **Use Case:** Extracting specific schema fields or array entries using `jq` syntax.


```python
from langchain_community.document_loaders import JSONLoader
```

```python
loader = JSONLoader(
    file_path="../docs/document_processing/sample.json",
    jq_schema=".Employees[]",
    text_content=False
)
docs = loader.load()
```

```python
print(f"Total Documents: {len(docs)}")
for i, doc in enumerate(docs):
    print(f"\n\nPage No: {i+1}")
    print(f"Source: {doc.metadata.get("source", "None")}")
    print(f"Content: {doc.page_content}")
```
