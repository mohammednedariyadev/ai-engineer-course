### Pandas

pandas builds directly on top of NumPy, adding explicit row indices, named columns, and flexible file I/O designed for tabular data.

**1. The Core Data Structures: Series vs. DataFrame**

Think of a Series as a single column with an index, and a DataFrame as an entire spreadsheet table composed of multiple Series sharing the same index.

```python
import pandas as pd
import numpy as np
```

```python
# Series: 1D labeled array
ages = pd.Series([25, 30, 35], index=["Alice", "Bob", "Charlie"], name="age")
print(ages)
```

```python
# Dataframe 2D Level Data structure (columns can have different types)
data = {
    "Age": [25, 30, 35],
    "City": ["New York", "London", "Tokyo"]
}
df = pd.DataFrame(data, index=["Alice", "Bob", "Charlie"])
print(df)
```

```python
# Extracting one column from a DataFrame returs a Series: df["Age"]
print(df["City"])
```

**2. Common Ways to Create DataFrames**

**From a Dictionary of Lists (Most Common)**

Keys become column headers; list items become row values.

```python
data = {
    "Name": ["Laptop", "Monitor", "Keyboard"],
    "Price": [1000, 25, 10],
    "In_Stock": [True, True, False]
}
df = pd.DataFrame(data)
print(df)
```

**From a List of Dictionaries (Row-Oriented)**

Useful when consuming API responses or JSON records where each item represents a row.

```python
records = [
    { "Name": "Laptop", "Price": 1000 },
    { "Name": "Monitor", "Price": 25 },
    { "Name": "Keyboard", "Price": 10 },
]
df = pd.DataFrame(records)
print(df)
```

**From a 2D NumPy Array**

Manually pass `columns` and optional `index` labels.

```python
arr = np.random.randint(10, 100, size=(3,2))
print("Array")
print(arr)

df = pd.DataFrame(arr, columns=["Score_A", "Score_B"])
print("\nPandas")
print(df)
```

**5.** `df["age"].fillna(df["age"].mean())` — **Mean Imputation**

This is a standard feature engineering pattern in machine learning called **mean imputation**. It replaces missing numeric values with the mathematical average of the non-missing values in that column.

How it evaluates:

1. `df["age"].mean()` calculates: $\frac{25.0 + 35.0}{2} = 30.0$ (ignoring the `NaN` s).
2. `.fillna(30.0)` substitutes every `NaN` in age with `30.0`.

```python
df["age"] = df["age"].fillna(df["age"].mean())
```

```python
df
```

*Tip: If data has extreme outliers, use `df["age"].median()` instead of `.mean()`. For categorical data (like city), use mode: `df["city"].fillna(df["city"].mode()[0])`.*

```python
# Employees table
df_emp = pd.DataFrame({
    "emp_id": [1, 2, 3, 4],
    "name": ["Alice", "Bob", "Charlie", "David"],
    "dept_id": [10, 20, 10, 99]  # Note: dept_id 99 doesn't exist in df_dept
})
print("Employee Data")
print(df_emp)

# Departments table
df_dept = pd.DataFrame({
    "dept_id": [10, 20, 30],       # Note: dept_id 30 has no employees
    "dept_name": ["Engineering", "HR", "Marketing"]
})
print("\n\nDepartment Data")
print(df_dept)
```

**1.** **The Four Join Types (`how=`)**

The `how` parameter dictates which rows are kept when keys do not match across both tables.

**Inner Join (`how="inner"`, default)**

Keeps only rows where the key exists in **both** DataFrames (intersection).

```python
pd.merge(df_emp, df_dept, on="dept_id", how="inner")
```

**Left Join (`how="left"`)**

Keeps **all rows from the left table** (`df_emp`). Missing matches from the right table become `NaN`.

```python
pd.merge(df_emp, df_dept, on="dept_id", how="left")
```

**Left Join (`how="left"`)**

Keeps all rows from the left table (`df_emp`). Missing matches from the right table become `NaN`.

```python
pd.merge(df_emp, df_dept, on="dept_id", how="right")
```

**Outer Join (`how="outer"`)**

Keeps **all rows from both tables** (union), filling non-matches with `NaN`.

```python
pd.merge(df_emp, df_dept, on="dept_id", how="outer")
```

**2. Handling Different Column Names:** `left_on` and `right_on`

When the key columns have different names in each table, map them explicitly:

```python
df_roles = pd.DataFrame({
    "id": [1, 2],
    "role": ["Dev", "Recruiter"]
})
df_roles
```

```python
# Merge employee table on 'emp_id' with roles table on 'id'
pd.merge(df_emp, df_roles, left_on="emp_id", right_on="id", how="inner")
```

**3. Handling Overlapping Non-Key Column Names:** `suffixes`

If both DataFrames share a column that isn't the merge key, Pandas appends `_x` and `_y` by default. Customize this with `suffixes`:

```python
df1 = pd.DataFrame({"id": [1], "location": ["NY"]})
df2 = pd.DataFrame({"id": [1], "location": ["London"]})

pd.merge(df1, df2, on="id", suffixes=("_hq", "_branch"))
```

**4. Merging on Index:** `left_index` and `right_index`

If a merge key resides in the index rather than a regular column:

```python
# Use the row index of df_dept instead of a column
df_dept_indexed = df_dept.set_index("dept_id")

pd.merge(df_emp, df_dept_indexed, left_on="dept_id", right_index=True, how="left")
```

**5. Tracking Match Origin:** `indicator=True`

Adds a special `_merge` column showing whether a record came from `both`, `left_only`, or `right_only`.

```python
merged = pd.merge(df_emp, df_dept, on="dept_id", how="outer", indicator=True)
print(merged[["emp_id", "dept_id", "_merge"]])
```

```python
df = pd.DataFrame({
    "Date": ["Mon", "Mon", "Tue", "Tue", "Mon"],
    "City": ["NYC", "LA",  "NYC", "LA",  "NYC"],
    "Temp": [70,    80,    72,    82,    68]  # Note: 2 entries for Mon-NYC (70 and 68)
})
df
```

**1.** `df.pivot()` — **Reshape Wide (No Aggregation)**

Turns unique values from one column into new column headers. It reshapes data purely on a 1-to-1 basis.

- Critical Rule: `pivot()` cannot handle duplicate index/column pairs. If combinations repeat, it raises a `ValueError: Index contains duplicate entries`.

Using a clean slice without duplicates:

```python
clearn_df = df.iloc[:4] # Rows 0 to 3 (unique combinations only)
clearn_df.pivot(index="Date", columns="City", values="Temp")
```

**2.** `pd.pivot_table()` — **Reshape Wide with Aggregation (Excel-Style)**

Functions like an Excel Pivot Table. It does everything `pivot()` does, but **safely handles duplicate entries** by aggregating them using `aggfunc` (defaults to `"mean"`).

**Handling Duplicates Automatically**

Because `df` has two readings for `Mon/NYC` (70 and 68), `pivot_table()` calculates their average (`69.0`):

```python
# aggfunc='mean' by default
df.pivot_table(index="Date", columns="City", values="Temp")
```

**Custom Aggregations and Grand Totals (margins)**

```python
df.pivot_table(
    index="Date",
    columns="City",
    values="Temp",
    aggfunc="max",
    margins=True,
    margins_name="Max_All"
)
```

**3.** `pd.melt()` — **Unpivot Wide to Long**

The exact inverse of pivoting. It takes multiple columns and unpivots them into row pairs (`variable` and `value`). This is crucial for Seaborn plotting and machine learning feature preparation.

```python
wide_df = pd.DataFrame({
    "Student": ["Alice", "Bob"],
    "Math": [90, 75],
    "English": [85, 92]
})
wide_df
```

**Melt it into Long Format:**

- `id_vars`: Columns to keep fixed as identifier variables.
- `value_vars`: Columns to unpivot (if omitted, melts all remaining columns).
- `var_name` / `value_name`: Names for the newly created columns.

```python
pd.melt(
    wide_df,
    id_vars=["Student"],
    value_vars=["Math", "English"],
    var_name="Subject",
    value_name="Score"
)
```

2. Ambiguous Dates: `format=` and `dayfirst=`

Different countries format dates differently (e.g., `04/09/2026` could mean April 9th or September 4th). Use `format` or `dayfirst` to prevent misinterpretation and speed up parsing.

**Using** `dayfirst=True` **(European / Indian format: DD/MM/YYYY)**

```python
s = pd.Series(["04/09/2026", "15/09/2026"])
s
```

```python
pd.to_datetime(s, dayfirst=True)
```

**Explicit Format String** (`format=`)

Explicitly specifying the structure bypasses heuristic guessing, running much faster on large datasets.

```python
date_strings = ["2026-Sep-18 09:30:00", "2026-Sep-19 14:15:00"]

# %Y = 4-digit year, %b = month abbrev, %d = day, %H = 24hr hour, %M = min, %S = sec
df = pd.DataFrame({
    "datetime": pd.to_datetime(date_strings, format="%Y-%b-%d %H:%M:%S")
})
df
```

Common format tokens:

- `%Y `(2026) vs `%y` (26)
- `%m` (numeric 09) vs `%b` (Sep) vs `%B` (September)
- `%d` (day of month: 01–31)
- `%H` (00–23) vs `%I` (01–12) with `%p` (AM/PM)

**3. Handling Dirty / Malformed Data** (`errors=`)

Real-world datasets often contain junk data like `"unknown"`, `"N/A"`, or invalid dates like `"2026-02-30"`.

- `errors='raise'` (default): Throws an exception and stops execution.
- `errors='coerce'`: Replaces invalid values with `NaT` (Not a Time, Pandas' null for dates).

```python
dirty_dates = ["2026-01-01", "not_a_date", "2026-02-31"]

pd.to_datetime(dirty_dates, errors="coerce")
```

**5. Why Convert? The** `.dt` **Accessor & Date Math**

Once converted to `datetime64`, access date components and calculate elapsed time directly:

```python
df = pd.DataFrame({"login_time": pd.to_datetime(["2026-09-18 14:30:00"])})

# Extract features
df["year"] = df["login_time"].dt.year                 # 2026
df["month_name"] = df["login_time"].dt.month_name()   # September
df["day_name"] = df["login_time"].dt.day_name()       # Friday
df["dayofweek"] = df["login_time"].dt.dayofweek       # 4 (0 is Monday)
df["hour"] = df["login_time"].dt.hour                 # 14

# Date arithmetic
now = pd.to_datetime("2026-09-20 14:30:00")
df["days_elapsed"] = (now - df["login_time"]).dt.days  # 2 days
df
```
