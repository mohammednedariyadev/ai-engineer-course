## Q21: Use `groupby()` and `.agg()` to find the minimum, maximum, and mean age, along with the total number of survivors, for each `Passenger_Class`.

```python
df.groupby(by=["Passenger_Class"]).agg({
    "Age": ["min", "max", "mean"],
    "Survived": "sum"
})
```

## Q22: Create a custom function that categorizes age into `"Child"` (under 18), `"Adult"` (18–60), and `"Senior"` (above 60). Apply this function to create a new column called `AgeGroup`.

```python
def categorize_age(age):
    if age < 18: return "Child"
    elif  age <= 60: return "Adult"
    else: return "Senior"

df["AgeGroup"] = df["Age"].apply(categorize_age)
df
```

## Q23: Find the names of the top 3 passengers who paid the absolute highest fares in 3rd class (`Passenger_Class == 3`).

```python
df[df["Passenger_Class"] == 3].nlargest(3, 'Fare')
```

## Q24: Extract the Surname (the text before the comma) from the `Name` column and store it in a new column called `Surname`.

```python
df["Surname"] = df["Name"].str.split(',').str[0]
df
```

## Q25: Use `pivot_table` to calculate the total number of survivors (`sum`) and total counts (`count`) cross-tabulated by `Passenger_Class` (rows) and `Gender` (columns).

```python
df.pivot_table(index='Passenger_Class', columns='Gender', values='Survived', aggfunc=['sum', 'count'])
```

## Q26: Identify the rows where a passenger's age is an outlier, defined here as being greater than 3 standard deviations from the dataset's mean age.

```python
mean_age = df["Age"].mean()
std_age = df["Age"].std()
outlier = df[df["Age"] > (mean_age + 3 * std_age)]
outlier
```

## Q27: Group the data by `Passenger_Class` and replace the missing `Age` values within each class using that specific class's mean age (using `.transform()`).

```python
df["Age"] = df.groupby("Passenger_Class")["Age"].transform(lambda x: x.fillna(x.mean()))
```
