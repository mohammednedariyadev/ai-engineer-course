
## 7. Structured Data (CSV, JSON, JSONL)

### CSV Files

* **Loader:** `CSVLoader`

* **Install:** No extra packages needed.

* **Use Case:** Each row is converted into an independent `Document` object where column headers serve as field labels.


```python
from langchain_community.document_loaders import CSVLoader
```

```python
loader = CSVLoader("../docs/document_processing/sample.csv")
docs = loader.load()
```

```python
print(f"Total Documents: {len(docs)}")
for i, doc in enumerate(docs[:2]):
    print(f"\n\nPage No: {i+1}")
    print(f"Source: {doc.metadata.get("source", "None")}")
    print(f"Content: {doc.page_content}")
```
