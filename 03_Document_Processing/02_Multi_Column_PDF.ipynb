
### 2. PDF with Precise Layout (Multi-column, Academic Papers)

* **Loader:** `PyMuPDFLoader` (or `PDFMinerLoader`)

* **Install:** `pip install pymupdf`

* **Use Case:** Multi-column magazines, newsletters, and papers where reading order matters and bounding-box coordinates may be needed.


```python
from langchain_community.document_loaders import PyMuPDFLoader
```

```python
# Good PDF loading
laoder = PyMuPDFLoader("../docs/document_processing/two-column.pdf")
docs = laoder.load()
```

```python
print(f"Total Documents {len(docs)}")
for i, doc in  enumerate(docs):
    print(f"\n\nPage No: {i+1}")
    print(f"Source: {doc.metadata.get("source", "None")}")
    print(f"Content: {doc.page_content[:40]}....{doc.page_content[-40:]}")
```
