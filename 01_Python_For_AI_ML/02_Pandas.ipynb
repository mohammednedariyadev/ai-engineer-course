### Pandas

pandas builds directly on top of NumPy, adding explicit row indices, named columns, and flexible file I/O designed for tabular data.

**1. The Core Data Structures: Series vs. DataFrame**

Think of a Series as a single column with an index, and a DataFrame as an entire spreadsheet table composed of multiple Series sharing the same index.

```python
import pandas as pd
import numpy as np
```

```python
# Series: 1D labeled array
ages = pd.Series([25, 30, 35], index=["Alice", "Bob", "Charlie"], name="age")
print(ages)
```

```python
# Dataframe 2D Level Data structure (columns can have different types)
data = {
    "Age": [25, 30, 35],
    "City": ["New York", "London", "Tokyo"]
}
df = pd.DataFrame(data, index=["Alice", "Bob", "Charlie"])
print(df)
```

```python
# Extracting one column from a DataFrame returs a Series: df["Age"]
print(df["City"])
```

**2. Common Ways to Create DataFrames**

**From a Dictionary of Lists (Most Common)**

Keys become column headers; list items become row values.

```python
data = {
    "Name": ["Laptop", "Monitor", "Keyboard"],
    "Price": [1000, 25, 10],
    "In_Stock": [True, True, False]
}
df = pd.DataFrame(data)
print(df)
```

**From a List of Dictionaries (Row-Oriented)**

Useful when consuming API responses or JSON records where each item represents a row.

```python
records = [
    { "Name": "Laptop", "Price": 1000 },
    { "Name": "Monitor", "Price": 25 },
    { "Name": "Keyboard", "Price": 10 },
]
df = pd.DataFrame(records)
print(df)
```

**From a 2D NumPy Array**

Manually pass `columns` and optional `index` labels.

```python
arr = np.random.randint(10, 100, size=(3,2))
print("Array")
print(arr)

df = pd.DataFrame(arr, columns=["Score_A", "Score_B"])
print("\nPandas")
print(df)
```

**3. Ingesting Data From Files and Databases**

Pandas includes built-in parsers starting with `pd.read_*` that handle types, missing values, and structure automatically.

**Reading CSV: `pd.read_csv()`**

The standard comma-delimited text format.

```python
# Basic read
df = pd.read_csv("../docs/pandas/tips.csv")

# Usefull Parameters
df = pd.read_csv(
    "../docs/pandas/tips.csv",
    sep=",",                            # Delimiter (e.g. '\t' for tap-separated)
    header=0,                           # Row number to use as a column name
    usecols=["total_bill", "tip"],      # Load onlt specific columns (save RAM)
    na_values=["?", "NA"]               # Treat custom symbols as NAN missing values
)
```

**Reading Excel: `pd.read_excel()`**

Reads `.xlsx` or `.xls` workbooks. Requires the `openpyxl` library installed.

```python
# Read the fist sheet by default
df = pd.read_excel('../docs/pandas/Sample-Sales-Data.xlsx')
print(df)
```

```python
# Specific a sheet name or index
df = pd.read_excel("../docs/pandas/Sample-Sales-Data.xlsx", sheet_name="Sheet1")
print(df)
```

```python
# Skip introductory title rows
df = pd.read_excel("../docs/pandas/Sample-Sales-Data.xlsx", skiprows=2)
print(df)
```

**Reading JSON: `pd.read_json()`**

Parses JSON strings or files into tabular format.

```python
# Direct json file ingestion
df = pd.read_json("../docs/pandas/world-population-by-country-2020.json")
print(df)

# Note: For deeply nested JSON, use json_normalize instead:
# import json
# with open("nested.json") as f:
#     data = json.load(f)
# df = pd.json_normalize(data, record_path=["items"], meta=["id"])
```

**Reading SQL: `pd.read_sql()`**

Executes an SQL query or reads a table directly via a database connection (typically created with SQLAlchemy).

```python
from sqlalchemy import create_engine

# 1. Establish the connection engine
engine = create_engine("sqlite:///company.db")
# Or PostgreSQL/MySQL: "postgresql://user:password@localhost:5432/mydb"

# 2. Query directly into a DataFrame
query = "SELECT user_id, email, signup_date FROM users WHERE active = 1"
df = pd.read_sql(query, con=engine)

# Alternatively, read an entire table without writing SQL
df_table = pd.read_sql_table("users", con=engine)
```

**9.** `df["column"]` — **Single Column as a Series**

Passing a **single string** extracts just that column as a 1D Pandas `Series`.

```python
countries = world_population["country"]

print("Data Type")
print(type(countries))

print("\ncountries")
print(world_population["country"])
```

**10.** `df[["col1", "col2"]]` — **Multiple Columns as a DataFrame**

Passing a list of column names (note the double brackets `[[...]]`) returns a new 2D `DataFrame`, preserving tabular structure.

*Tip: Even for a single column, `df[["City"]]` forces the output to remain a DataFrame rather than decaying into a Series.*

```python
subset = world_population[["country", "world_share"]]

print("Datatype")
print(type(subset))

print("\nSubset")
print(subset)
```

**11.** `df.loc[]` — **Label-Based Selection:** `[row_label, col_label]`

`loc` works exclusively with **explicit index and column names.**

- **Rule**: When slicing with `loc` (`"r1":"r3"`), the endpoint is inclusive.

```python
# Select a single cell:
world_population.loc[0, "country"]
```

```python
# Select rows by label:
world_population.loc[0]
```

```python
# Row slice: both 'r0' and 'r2' are INCLUDED
world_population.loc[0 : 5]
```

```python
# Select specific rows AND columns:
world_population.loc[0: 10, ["country", "world_share"]]
```

```python
# Conditional row filtering with loc:
world_population.loc[world_population["position"] > 10, ["country", "world_share"]]
```

```python
world_population.loc[(world_population["position"] > 10) & (world_population["position"] < 20), ["country", "world_share"]]
```

**12.** `df.iloc[]` — **Integer Position-Based Selection:** `[row_idx, col_idx]`

`iloc` (integer location) completely ignores index labels and column names, relying strictly on **zero-based numeric coordinates ($0, 1, 2, \dots$)**, identical to NumPy.

- **Rule**: Slicing with `iloc` follows standard Python rules: the end index is **exclusive**.

```python
print(world_population.iloc[0:2, 0:2])
```

**13.** `isin()` — **Multiple OR Conditions for Values**

Instead of writing 

`(df["City"] == "Paris") | (df["City"] == "Rome") | (df["City"] == "Berlin")`, 

pass an iterable (list, set, tuple) to `isin()`.

- It generates a **boolean mask** (`True` if the element matches anything in the list).

```python
# Select rows where city is either India, China and Indonesia
cities = ["India", "China", "Indonesia"]
world_population[world_population["country"].isin(cities)]
```

**14.** Negation (NOT IN)

Prefix the condition with `~` to exclude items:

```python
# Select rows where city is NOT in India, China and Indonesia
cities = ["India", "China", "Indonesia"]
world_population[~world_population["country"].isin(cities)]
```

**15.** `between()` — **Range Filtering (Lower & Upper Bound)**

Instead of writing `(df["Age"] >= 25) & (df["Age"] <= 35)`, `between()` checks whether numeric or datetime values fall inside a range.

- **Syntax**: `df["col"].between(left, right, inclusive="both")`
- `inclusive` options: `"both"` (default), `"neither"`, `"left"`, or `"right"`.

```python
# Filter position between 50 and 60 (inclusive)
world_population[world_population["position"].between(50, 60)]
```

```python
# Exclusive Range Check
# Position > 4 AND Position < 8
world_population[world_population["position"].between(4,8, inclusive="neither")]
```

**16.** `query()` — **SQL-Style String Expressions**

`df.query()` evaluates an expression string. It eliminates repetitive references like `df[...]` and handles complex logical expressions cleanly.

**Basic Comparisons and Multiple Conditions**

Inside the query string, use standard words like `and`, `or`, `in`, and `not`.

```python
world_population.query("country == 'India'")
```

```python
world_population.query("country in ['India', 'China']")
```

```python
country = "India"
world_population.query("country == @country")
```
